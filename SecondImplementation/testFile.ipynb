```python
from ai_interface import run_headless_battle
import json

def my_policy(state, battle, boss_slots):
    turn = state["turn_number"]
    newState = state.copy()
    commands = []

    del newState["turn_number"]

    if turn == 1:
        print(json.dumps(newState, indent= 4))


    
    for i, unit in enumerate(state["units"]):
        if unit["is_dead"] or unit[ "is_staggered"]:
            continue  # needs no command at all
        
        relevantState = newState["units"][i]
        if(turn == 1):
            print(json.dumps(relevantState, indent=4))

        hp = unit["hp"]

        isDead = int(unit["is_dead"])
        isStaggered = int(unit["is_staggered"])
        staggersRemaining = unit["num_stagger_thresholds_remaining"]
        burnPotency = unit["burn_potency"]
        burnCount = unit["burn_count"]
        tremorPotency = unit["tremor_potency"]
        tremorCount = unit["tremor_count"]
        magicBullets = unit["magic_bullets"]
        clashPower = unit["clash_power_level"]

        damageReduction = int(unit["damage_reduction_active"])

        skill0, skill1, skill2, skill3 = unit["skills"]

        unitNumber = i

        parameters = [hp, isDead, isStaggered, staggersRemaining, burnPotency, burnCount, tremorPotency, tremorCount, magicBullets, clashPower, damageReduction, skill0, skill1, skill2, skill3, unitNumber]

        for j, otherUnit in enumerate(state["units"]):
            if i == j:
                continue
        
            otherUnitIsDead = otherUnit["is_dead"]
            otherUnitIsStaggered = otherUnit["is_staggered"]
            otherUnitStaggersRemaining = otherUnit["num_stagger_thresholds_remaining"]
            otherUnitClashPower = otherUnit["clash_power_level"]
            otherUnitNumber = j
            otherSkill0, otherSkill1, otherSkill2, otherSkill3 = otherUnit["skills"]

            parameters += [otherUnitIsDead, otherUnitIsStaggered, otherUnitStaggersRemaining, otherUnitClashPower, otherUnitNumber, otherSkill0, otherSkill1, otherSkill2, otherSkill3]

            
        
        
        commands.append([0, 0])  # bottom skill, attack unopposed
    return commands

result = run_headless_battle(my_policy, seed=42)
print(result["outcome"], result["turns"])
```
